```python
A = np.array([[4, 7],
              [2, 6]])

det = np.linalg.det(A)
inv = np.linalg.inv(A)

print("Визначник:", det)
print("Обернена матриця:")
print(inv)
print("Перевірка:")
print(A @ inv)
```

```python
sales = np.array([[120, 150, 170, 160],
                  [80, 90, 100, 110],
                  [200, 180, 210, 230]])

product_totals = np.sum(sales, axis=1)
month_totals = np.sum(sales, axis=0)

print("Продажі товарів:", product_totals)
print("Продажі по місяцях:", month_totals)
print("Найкращий товар:", np.argmax(product_totals) + 1)
print("Найкращий місяць:", np.argmax(month_totals) + 1)
print("Середні продажі товарів:", np.mean(sales, axis=1))
```

```python
A = np.random.randint(1, 21, (5, 5))
print(A)

print("Головна діагональ:", np.trace(A))
print("Побічна діагональ:", np.trace(np.fliplr(A)))
print("Максимум по рядках:", np.max(A, axis=1))
print("Мінімум по стовпцях:", np.min(A, axis=0))

mean = np.mean(A)
B = A.copy()
B[B < mean] = 0

print("Середнє:", mean)
print(B)
```

```python
X = np.array([[2, 4, 6],
              [1, 3, 5],
              [5, 7, 9]])

print("1-2:", np.linalg.norm(X[0] - X[1]))
print("1-3:", np.linalg.norm(X[0] - X[2]))
print("2-3:", np.linalg.norm(X[1] - X[2]))
```

```python
A = np.array([[1, 2, 3, 4],
              [2, 3, 4, 5],
              [3, 4, 5, 6],
              [4, 5, 6, 7]], dtype=float)

norms = np.linalg.norm(A, axis=1)
B = A / norms[:, np.newaxis]

print("Норми:", norms)
print(B)
```

```python
A = np.array([[1, 2, 1],
              [2, 1, -1],
              [3, -1, 2]])

b = np.array([9, 8, 3])

x = np.linalg.solve(A, b)
check = A @ x

print("Розв'язок:", x)
print("Перевірка:", check)
print("b:", b)
print(np.allclose(check, b))
```
